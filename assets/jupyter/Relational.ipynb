Sales Performance by Country

Displaying the total sales for each country is a direct and effective way to identify top-performing geographic markets. It provides a clear, quantitative measure of where the business is generating the most revenue.

```python
print("Total sales per country:")
print("-" * 50)

# Query to calculate total sales per country
query_sales_by_country = """
    SELECT
        c.country,
        SUM(od.quantityOrdered * od.priceEach) AS totalSales
    FROM
        customers c
    JOIN
        orders o ON c.customerNumber = o.customerNumber
    JOIN
        orderdetails od ON o.orderNumber = od.orderNumber
    GROUP BY
        c.country
    ORDER BY
        totalSales DESC;
"""

df_sales_by_country = pd.read_sql(query_sales_by_country, conn)
print(df_sales_by_country)
```

Customer Distribution

A list of all customers and their assigned sales representatives. It uses a JOIN to link the *customers* and *employees* tables, providing a clear view of the workload distribution among the sales team. 

```python
print("Customer and their sales representatives:")
print("-" * 50)

# Query to retrieve customer names and their sales representatives' names
query_customer_sales_rep = """
    SELECT
        c.customerName,
        e.firstName || ' ' || e.lastName AS salesRepName
    FROM
        customers c
    JOIN
        employees e ON c.salesRepEmployeeNumber = e.employeeNumber
    ORDER BY
        salesRepName, c.customerName;
"""

df_customer_sales_rep = pd.read_sql(query_customer_sales_rep, conn)
print(df_customer_sales_rep)
```

Order Size

Size of each order by counting the number of unique products it contains. 

```python
print("Unique products in each order:")
print("-" * 50)

# Query to get the number of unique products in each order
query_order_size = """
    SELECT
        orderNumber,
        COUNT(productCode) AS uniqueProductsCount
    FROM
        orderdetails
    GROUP BY
        orderNumber
    ORDER BY
        uniqueProductsCount DESC;
"""

df_order_size = pd.read_sql(query_order_size, conn)
print(df_order_size)
```

Relationships

Different types of joins provide distinct insights into the structure of the sales force and the company's internal organization.

```python
print("Relationships")
print("-" * 50)

# ----- INNER JOIN -----
# Shows only customers who have a sales representative
print("--- Customers with an assigned sales representative ---")
query_inner_join = """
    SELECT
        c.customerName,
        e.firstName || ' ' || e.lastName AS salesRepName
    FROM
        customers c
    JOIN
        employees e ON c.salesRepEmployeeNumber = e.employeeNumber;
"""
df_inner = pd.read_sql(query_inner_join, conn)
print(df_inner)
print("\n" + "="*80 + "\n")

# ----- LEFT JOIN -----
# Shows ALL customers, even if they do not have a sales representative
print("--- All customers, including those without a representative ---")
query_left_join = """
    SELECT
        c.customerName,
        e.firstName || ' ' || e.lastName AS salesRepName
    FROM
        customers c
    LEFT JOIN
        employees e ON c.salesRepEmployeeNumber = e.employeeNumber;
"""
df_left = pd.read_sql(query_left_join, conn)
print(df_left)
print("\n" + "="*80 + "\n")

# ----- SELF-JOIN -----
# Shows the employee hierarchy: who reports to whom
print("--- Employee hierarchy ---")
query_self_join = """
    SELECT
        e.firstName || ' ' || e.lastName AS employeeName,
        m.firstName || ' ' || m.lastName AS managerName
    FROM
        employees e
    JOIN
        employees m ON e.reportsTo = m.employeeNumber
    ORDER BY
        managerName;
"""
df_self = pd.read_sql(query_self_join, conn)
print(df_self)
print("\n" + "="*80 + "\n")
```
